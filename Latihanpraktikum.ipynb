### Tugas Analisis 1:
Ketika mencoba menggunakan list Python biasa, program akan error dan menampilkan kesalahan TypeError: can't multiply sequence by non int of type 'float'. Tapi jika menggunakan Array program akan mengalikan tiap angka di dalam array, tidak di jadikan satu


```python
#1. Operasi dasar Numpy Array
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga Terendah :', harga.min())
print('Harga setelah diskon 10%:', harga * 0.9)
```

### Tugas Analisis 4:
Mengapa pada contoh di bawah kolom terjual diisi (fillna) sedangkan baris
dengan menu kosong justru dihapus (dropna)? Diskusikan alasan logisnya.

```python
#4. Menangani Missing Value
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong

print(df.head())
```

### Jawaban 4:
Data "Terjual" di isi 0 karena angka Nol dan non-null bisa diartikan sebagai tidak ada penjualan. Sedangkan baris dengan menu kosong dihapus karena transaksi tanpa identitas produk merusak akurasi data

### Tugas Analisis 5:
Setelah drop_duplicates(), berapa jumlah baris dataset sekarang dibanding
sebelumnya? Mengapa penting memastikan tipe data (dtypes) sudah benar sebelum data
dianalisis lebih lanjut?

```python
#5. Menangani Duplikat dan Tipe Data
print(df.duplicated().sum()) # jumlah baris duplikat
print(df.info())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)

print(df.info())
```

### Tugas Analisis 6
Dari hasil groupby di bawah, menu apa yang menghasilkan total_pendapatan
tertinggi? Bagaimana informasi ini bisa membantu pengambilan keputusan di kantin sekolah?

```python
#Data Manipulation (Filtering, Sorting, Groupby)
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
